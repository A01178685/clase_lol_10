```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = 'antropometria-dataset.csv'
df = pd.read_csv(file_path)
print('Filas y columnas:', df.shape)
df.head()
```

```python
variables = ['peso', 'cintura', 'cadera']

fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
for eje, variable in zip(ejes, variables):
    df[variable].dropna().plot(kind='hist', bins=25, ax=eje)
    eje.set_title('Histograma de ' + variable)
    eje.set_xlabel(variable)
    eje.set_ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```

```python
fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
for eje, variable in zip(ejes, variables):
    eje.boxplot(df[variable].dropna())
    eje.set_title('Box plot de ' + variable)
    eje.set_ylabel(variable)
plt.tight_layout()
plt.show()
```

```python
promedios_edad = df.groupby('edad')[variables].mean()

fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
for eje, variable in zip(ejes, variables):
    promedios_edad[variable].plot(kind='line', ax=eje)
    eje.set_title('Promedio de ' + variable + ' por edad')
    eje.set_xlabel('Edad (años)')
    eje.set_ylabel(variable)
plt.tight_layout()
plt.show()
```

```python
df.plot(kind='scatter', x='cintura', y='peso',
        figsize=(8, 5), alpha=0.25, s=8)
plt.title('Peso y cintura: todos los registros')
plt.xlabel('Cintura')
plt.ylabel('Peso')
plt.show()
```

```python
for sexo, nombre in [(1, 'Hombres'), (2, 'Mujeres')]:
    grupo = df[df['sexo'] == sexo]
    print(nombre, '- registros:', len(grupo))

    fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
    for eje, variable in zip(ejes, variables):
        grupo[variable].dropna().plot(kind='hist', bins=25, ax=eje)
        eje.set_title(nombre + ': ' + variable)
        eje.set_xlabel(variable)
        eje.set_ylabel('Frecuencia')
    plt.tight_layout()
    plt.show()

    fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
    for eje, variable in zip(ejes, variables):
        eje.boxplot(grupo[variable].dropna())
        eje.set_title(nombre + ': ' + variable)
        eje.set_ylabel(variable)
    plt.tight_layout()
    plt.show()

    promedios = grupo.groupby('edad')[variables].mean()
    fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
    for eje, variable in zip(ejes, variables):
        promedios[variable].plot(kind='line', ax=eje)
        eje.set_title(nombre + ': promedio por edad de ' + variable)
        eje.set_xlabel('Edad (años)')
        eje.set_ylabel(variable)
    plt.tight_layout()
    plt.show()

    grupo.plot(kind='scatter', x='cintura', y='peso',
               figsize=(8, 5), alpha=0.25, s=8)
    plt.title(nombre + ': peso y cintura')
    plt.xlabel('Cintura')
    plt.ylabel('Peso')
    plt.show()
```

```python
for sexo, nombre in [(0, 'Todos'), (1, 'Hombres'), (2, 'Mujeres')]:
    grupo = df if sexo == 0 else df[df['sexo'] == sexo]
    print(nombre)
    print('Asimetría:')
    print(grupo[variables].skew().round(2))
    print('Correlación peso-cintura:',
          round(grupo[['peso', 'cintura']].corr().loc['peso', 'cintura'], 3))
    print('Medianas:')
    print(grupo[variables].median().round(2))
```
