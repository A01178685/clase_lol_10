```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model

file_path = 'antropometria-dataset.csv'
df = pd.read_csv(file_path)
df.head()
```

```python
datos = df[['peso', 'cintura']].dropna().sort_values('cintura').copy()
x = datos[['cintura']]
y = datos['peso']

lm = linear_model.LinearRegression()
modelo = lm.fit(x, y)
datos['peso_estimado'] = modelo.predict(x)

print('Intercept (B0):', modelo.intercept_)
print('Slope / coefficient (B1):', modelo.coef_[0])
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(datos['cintura'], datos['peso'], alpha=0.25, s=8, label='Actual peso')
plt.plot(datos['cintura'], datos['peso_estimado'], color='red',
         label='Estimated peso')
plt.xlabel('Cintura')
plt.ylabel('Peso')
plt.title('Peso versus cintura: all participants')
plt.legend()
plt.show()
```

```python
datos['residuo'] = datos['peso'] - datos['peso_estimado']
print('Residual = actual peso - estimated peso')
print(datos[['cintura', 'peso', 'peso_estimado', 'residuo']].head())
```

```python
for sexo, nombre in [(1, 'Men'), (2, 'Women')]:
    grupo = df[df['sexo'] == sexo][['peso', 'cintura']].dropna().sort_values('cintura').copy()
    x_grupo = grupo[['cintura']]
    y_grupo = grupo['peso']

    modelo_grupo = linear_model.LinearRegression()
    modelo_grupo.fit(x_grupo, y_grupo)
    grupo['peso_estimado'] = modelo_grupo.predict(x_grupo)
    grupo['residuo'] = grupo['peso'] - grupo['peso_estimado']

    print(nombre, '- observations:', len(grupo))
    print('Intercept (B0):', modelo_grupo.intercept_)
    print('Slope / coefficient (B1):', modelo_grupo.coef_[0])
    print(grupo[['cintura', 'peso', 'peso_estimado', 'residuo']].head())

    plt.figure(figsize=(8, 5))
    plt.scatter(grupo['cintura'], grupo['peso'], alpha=0.25, s=8,
                label='Actual peso')
    plt.plot(grupo['cintura'], grupo['peso_estimado'], color='red',
             label='Estimated peso')
    plt.xlabel('Cintura')
    plt.ylabel('Peso')
    plt.title('Peso versus cintura: ' + nombre)
    plt.legend()
    plt.show()
```
