```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = 'antropometria-dataset.csv'
df = pd.read_csv(file_path)
df.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['sistol'].dropna().plot(kind='hist', bins=25, ax=axes[0])
df['diastol'].dropna().plot(kind='hist', bins=25, ax=axes[1])
axes[0].set_title('Systolic pressure (sistol)')
axes[1].set_title('Diastolic pressure (diastol)')
for ax in axes:
    ax.set_xlabel('Blood pressure')
    ax.set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['peso'].dropna().plot(kind='hist', bins=25, ax=axes[0])
df['cintura'].dropna().plot(kind='hist', bins=25, ax=axes[1])
axes[0].set_title('Weight (peso)')
axes[1].set_title('Waist (cintura)')
axes[0].set_xlabel('Weight')
axes[1].set_xlabel('Waist')
for ax in axes:
    ax.set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
df.plot(kind='scatter', x='cintura', y='peso',
        figsize=(8, 5), alpha=0.25, s=8)
plt.title('Weight versus waist: all participants')
plt.xlabel('Waist (cintura)')
plt.ylabel('Weight (peso)')
plt.show()
```

```python
for sex, label in [(1, 'Men'), (2, 'Women')]:
    group = df[df['sexo'] == sex]
    print(label, '- participants:', len(group))

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    group['sistol'].dropna().plot(kind='hist', bins=25, ax=axes[0])
    group['diastol'].dropna().plot(kind='hist', bins=25, ax=axes[1])
    axes[0].set_title(label + ': systolic pressure')
    axes[1].set_title(label + ': diastolic pressure')
    for ax in axes:
        ax.set_xlabel('Blood pressure')
        ax.set_ylabel('Frequency')
    plt.tight_layout()
    plt.show()

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    group['peso'].dropna().plot(kind='hist', bins=25, ax=axes[0])
    group['cintura'].dropna().plot(kind='hist', bins=25, ax=axes[1])
    axes[0].set_title(label + ': weight')
    axes[1].set_title(label + ': waist')
    axes[0].set_xlabel('Weight')
    axes[1].set_xlabel('Waist')
    for ax in axes:
        ax.set_ylabel('Frequency')
    plt.tight_layout()
    plt.show()

    group.plot(kind='scatter', x='cintura', y='peso',
               figsize=(8, 5), alpha=0.25, s=8)
    plt.title(label + ': weight versus waist')
    plt.xlabel('Waist (cintura)')
    plt.ylabel('Weight (peso)')
    plt.show()
```

```python
variables = ['sistol', 'diastol', 'peso', 'cintura']
for label, group in [('All', df),
                     ('Men', df[df['sexo'] == 1]),
                     ('Women', df[df['sexo'] == 2])]:
    print(label)
    print('Skewness:')
    print(group[variables].skew().round(2))
    print('Weight-waist correlation:',
          round(group[['peso', 'cintura']].corr().loc['peso', 'cintura'], 3))
```
